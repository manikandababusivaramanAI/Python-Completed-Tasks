### 3. Write a program that handles division by zero  using try-except

```python
num = int(input ("Enter the numerator:"))
den = int(input ("Enter the denominator:"))

result = num/den

print (result)
```

```python
num = int(input ("Enter the numerator:"))
den = int(input ("Enter the denominator:"))

try:
    divison = num/den
    print (divison)

except ZeroDivisionError:
        print ("Enter the number greater then zero!")
        print ("Please Try Again!")
```

### 4. Given a String, count how many vowels it has?

```python
vowels = ["a","e","i","o","u"]

words = []

user_input = str(input("Enter any english word:"))

for i in user_input.lower():
    if i in vowels:
        words.append(i)
print(words)
print ("Vowels count in given word:",len(words))
```
