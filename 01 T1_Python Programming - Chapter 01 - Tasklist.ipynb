```python
### 1. length Function - it returns count of characters in the given string
maria = len("maria")
mani = len("manikandababu")
print (maria)
print (mani)
```

```python
### 6. isalnum() method - Returns "TRUE" if the given string in alphanumeric

num1="mani1234".isalnum()
print(num1)
num2="Rajini&Kamal".isalnum()
print (num2)
num3="Rajini Kamal".isalnum()
print (num3)
```

```python
### 7. isaplpha() method - Returns "TRUE" if the given string contains only alphabets
num1="manikandababu".isalpha()
print(num1)
num2="manikandababu123".isalpha()
print(num2)
num3="Rajini Kamal".isalpha()
print (num3)
```

```python
### 8. isdigit() method - Returns "TRUE" if the all the characters in the given string contains only numbers
num1="1234".isdigit()
num2="mani5678".isdigit()
print (num1,num2)
```

```python
### 9. islower() method - Returns "TRUE" if the all the characters in the given string are lowercase
num1="manikandababu".islower()
num2="Ramesh".islower()
print (num1,num2)
```

```python
### 10. isspace() method - Returns "TRUE", if there are only whitespace characters in the string
num1="   ".isspace()
num2="".isspace()
print (num1,num2)
```

```python
### 11. isupper() method - Returns "TRUE", if all the given characters in the string as Uppercase
num1="MANIKANDABABU".isupper()
num2="Manikandababu".isupper()
print(num1,num2)
```

```python
### 12. lower() method - converts the copy of a given string to a lowercase
num1 = "MANIKANDABABU".lower()
num2 = "ProITBridgE".lower()
print (num1)
print (num2)
```

```python
### 13. Upper() method - converts the copy of a given string to a Uppercase
num1 = "manikandababu".upper()
num2 = "proitbridge".upper()
print (num1)
print (num2)
```

```python
### 14. Strip method - Removes the spaces from Right and left side of the given string
num1 = "  mani   ".strip()
num2 = "babu   ".strip()
num3 = "mani babu".strip()
print (num1)
print (num2)
print (num3)
```

```python
### 15. Title method - return "TRUE", if the first letter in capital and other letters in lowercase.
num1="Manikandababu".istitle()
num2="MANIKANDABABU".istitle()
print (num1)
print (num2)
```

```python
### 16. Replace method - Returns a copy of a string with replace old value with new value
num1="i like india".replace('like','love')
num2 = "hello world".replace('world','python')
print (num1)
print (num2)
```

```python
### 17. join method -joins a string
join1='$'.join("manikandababu")
join2='##'.join(["mani","babu"])
print (join1)
print (join2)
```

```python
### 18. startswith() method - Returns "True", if the given string starts with substring.
name1="manikandababu".startswith('mani')
name2="manikandababu".startswith('babu')
print (name1)
print (name2)
```

```python
### 19. endsswith() method - Returns "True", if the given string starts with substring.
name1="manikandababu".endswith('babu')
name2="manikandababu".endswith('mani')
print (name1)
print (name2)
```

```python
### 20.Split method - it splits the string based on given substring but output returns as list
name1="I love india".split()
name2="I love india".split('o')
print (name1)
print (name2)
```

```python
### 21.partition method - it splits the string based on given substring but output returns as tuple
name1="I love india".partition('love')
print (name1)
name2="I love india".partition('i')
print (name2)
```

## TASK 3 - Rock, Paper, Scissors Game

```python
print ("Welcome to Rock, Paper, Scissors Game")
print ("Enter any one of the 03 choices - Rock/Paper/Scissors:")

secret_input = ["Rock","Paper","Scissors"]

import random

while True:

    for i in range(0,3):
            secret_input = random.choice(secret_input)
                
            user_inputs = input("Enter your choics:").capitalize()

            if user_inputs not in secret_input:
                print ("Invalid Entry! please enter as per instructions!")
                
            if secret_input == user_inputs:
                print ("You Wins!")
                break
            else:
                print ("You Lose!")        
             
    print ("Do you want to Play Again, Type YES or NO:")
    re_play = input ("Enter YES or NO:")
        
    if re_play.upper() == "YES":
        continue
    elif re_play.upper() == "NO":
        print ("Thanks for Playing")
        break
```

## TASK 4 - Word Guessing Game (Mini Hangman) - Completed using Chatgpt only

```python
print("Welcome to Word Guessing Game!")

import random

words = ["python", "machine", "learning", "bridge"]

word = random.choice(words)
guessed = []
attempts = 6

while attempts > 0:
    display = ""

    for letter in word:
        if letter in guessed:
            display += letter + " "
        else:
            display += "_ "

    print("\nWord:", display)
    print("Attempts left:", attempts)

    if "_" not in display:
        print("Congratulations! You guessed the word!")
        break

    guess = input("Guess one letter: ").lower()

    if len(guess) != 1 or not guess.isalpha():
        print("Please enter only one letter.")
        continue

    if guess in guessed:
        print("You already guessed that letter!")
        continue

    guessed.append(guess)

    if guess in word:
        print("Correct guess!")
    else:
        print("Wrong guess!")
        attempts -= 1

if attempts == 0:
    print("\nGame Over! The word was:", word)    
```
